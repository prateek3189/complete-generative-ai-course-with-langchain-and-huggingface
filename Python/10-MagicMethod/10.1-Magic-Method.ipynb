# Magic Method

```python
class Person:
  pass

person = Person()
```

```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def __str__(self):
    return f"This is custom method: {self.name}, {self.age}"

person = Person("Prateek", 37)
print(person)
```
