```python
## Modeling a Bank Account
class Account:
  def __init__(self, owner, balance = 0):
    self.owner = owner
    self.balance = balance

  def deposit(self, amount):
    self.balance += amount
    print(f"{amount} is deposited. New balance is {self.balance}") 

  def withdraw(self, amount):
      if amount > self.balance:
         print("Amount is not enough to withdraw")
      else:
        self.balance -= amount
        print(f"{amount} is withdrawn. New balance is {self.balance}") 
  def getBalance(self):
     print(f"Current Balance is: {self.balance}")

hdfc = Account("Prateek", 5000)
hdfc.deposit(344000)
hdfc.withdraw(10000)
hdfc.getBalance()
```

```python
## Inheritence

class Vehicle:
  def __init__(self, name, sitting):
    self.name = name
    self.sitting = sitting

  def drive(self):
    print(f"I am driving {self.name}")

class Bmw(Vehicle):
  def __init__(self, name, sitting, engineType):
    super().__init__(name, sitting)
    self.engineType = engineType

  def getEngineType(self):
    print(f"I have {self.name} that have a capacity of {self.sitting} and engin type is {self.engineType}")



# car1 = Vehicle("Prateek", 4)
# car1.drive()

bmw = Bmw("Prateek", 4, "Auto")
bmw.drive()
bmw.getEngineType()
```

```python
## Multiple Inheritence

class Animal:
  def __init__(self, name):
    self.name = name

  def speak(self):
    print("Child classes must implement this method")

class Pet:
  def __init__(self, owner):
    self.owner = owner

class Dog(Animal, Pet):
  def __init__(self, name, owner):
    Animal.__init__(self, name)
    Pet.__init__(self, owner)

  def speak(self):
      return f"Name is {self.name} and the owner is {self.owner}"

butcus = Dog("Butcus", "Rocky")
butcus.speak()

# animal = Animal("Butcus")
# animal.speak()
```
