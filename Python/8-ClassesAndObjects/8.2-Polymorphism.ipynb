# Polymorphism

```python
class Animal:
  def speak(self):
    return "Sound of the animal"

class Dog(Animal):
  def speak(self):
    return "Dog make a sound Wooof!!"

class Cat(Animal):
  def speak(self):
    return "Cat make a sound Meaww!!"

def animal_speak(animal):
  print(animal.speak())

dog = Dog()
# print(dog.speak())

cat = Cat()
# print(cat.speak())

animal_speak(dog)
animal_speak(cat)
```

```python
class Shape:
  def area(self):
    return "The Area of the figure"

class Rectangle(Shape):
  def __init__(self, width, height):
    self.width = width
    self.height = height

  def area(self):
    return self.width * self.height

class Circle(Shape):
  def __init__(self, radius):
    self.radius = radius

  def area(self):
    return 3.14 * self.radius * self.radius

# Function for Polymorphism
def print_area(shape):
  print(f" The are is {shape.area()}")

ract = Rectangle(10, 20)
cir = Circle(23)

print_area(ract)
print_area(cir)
  
```

```python
# Abstract
from abc import ABC, abstractmethod

## Define abstract class
class Vehicle(ABC):
  @abstractmethod
  def start_engine(self):
    pass

class Car(Vehicle):
  def start_engine(self):
      return "Car Started"

class MoterCycle(Vehicle):
  def start_engine(self):
    return "Bike started"

def start_engine(vehicle):
  print(vehicle.start_engine())

car = Car()
bike = MoterCycle()

start_engine(car)
start_engine(bike)
```
